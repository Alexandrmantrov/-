```python
import easyocr
import telebot
from telebot import types
import requests
```

```python
def ocr(file):
  reader = easyocr.Reader(['ru'])
  result = reader.readtext(file, detail = 0)
  spis =' '.join(result)
  return spis
```

```python
def document (file):
    with open("file.txt", "w") as output:
      output.write(file)
    return output
```
